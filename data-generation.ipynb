```python
import json
import pandas as pd

# Load in from libs/ImageReward/benchmark/benchmark-prompts.json
with open("libs/ImageReward/benchmark/benchmark-prompts.json", "r") as f:
    prompts = json.load(f)

dataset = pd.DataFrame([
    {
        "prompt_id": prompt["id"], 
        "prompt": prompt["prompt"], 
        "rubric_type": "ImageReward",
        "rubric_prompt": prompt["prompt"]
    }
    for prompt in prompts
])

print(dataset)
```

```python
# Split into dev and test, fixing the seed for reproducibility
SEED = 42
dev_dataset = dataset.sample(frac=0.7, random_state=SEED)
test_dataset = dataset.drop(dev_dataset.index)

# Save to csv
dev_dataset.to_csv("data/dev_dataset.csv", index=False)
test_dataset.to_csv("data/test_dataset.csv", index=False)
```
