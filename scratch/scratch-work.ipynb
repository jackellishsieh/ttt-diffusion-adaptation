```python
from diffusers import StableDiffusionPipeline
import torch
from PIL import Image
import matplotlib.pyplot as plt

# Load the pretrained Stable Diffusion 2.1 model
# Using FP16 for lower VRAM usage (if supported by GPU)
model_id = "stabilityai/stable-diffusion-2-1"

pipe = StableDiffusionPipeline.from_pretrained(
    model_id,
    torch_dtype=torch.float16,
    safety_checker=None,    # disable NSFW checker for simplicity
).to("cuda")
```

```python
prompt = "a cozy mountain cabin at sunset, high quality, detailed, digital art"

# --- Step 5: Generate an image ---
image = pipe(prompt).images[0]

# --- Step 6: Display the result ---
plt.imshow(image)
plt.axis("off")
plt.show()

# --- Step 7 (optional): Save it ---
image.save("sample_sd21.png")
```
